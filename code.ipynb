# 1. Vanishing and Exploding Gradient Problem

## Vanishing Gradient Problem
The vanishing gradient problem occurs when gradients become extremely small during backpropagation in deep neural networks. Because of this, weights are updated very slowly and the model fails to learn effectively.

### Causes
- Deep neural networks
- Sigmoid or tanh activation functions
- Improper weight initialization

### Effects
- Slow training
- Poor convergence
- Lower accuracy

### Solutions
- ReLU activation function
- Batch Normalization
- Residual Networks (ResNet)
- Xavier/He initialization

---

## Exploding Gradient Problem
The exploding gradient problem occurs when gradients become excessively large during training, leading to unstable learning.

### Effects
- Large weight updates
- NaN loss values
- Unstable model

### Solutions
- Gradient Clipping
- Proper initialization
- Batch Normalization
- Smaller learning rate



# 2. Bank Customer Churn Prediction using ANN

## Objective
To predict whether a customer will leave the bank using an Artificial Neural Network (ANN).

## Steps Involved
1. Import Libraries
2. Load Dataset
3. Perform EDA
4. Data Preprocessing
5. Build ANN Model
6. Train Model
7. Evaluate Model


```

# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```

# Load Dataset
# Replace file path according to your system
df = pd.read_csv('Churn_Modelling.csv')

# Display first 5 rows
df.head()
```

```

# Basic Information
df.info()

# Statistical Summary
df.describe()
```

```

# Check Null Values
df.isnull().sum()
```

```

# Data Visualization
sns.countplot(x='Exited', data=df)
plt.title('Customer Churn Distribution')
plt.show()
```

```

# Data Preprocessing

# Remove unnecessary columns
df = df.drop(['RowNumber', 'CustomerId', 'Surname'], axis=1)

# Encode Gender
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])

# One Hot Encoding for Geography
df = pd.get_dummies(df, columns=['Geography'], drop_first=True)

# Split Features and Target
X = df.drop('Exited', axis=1)
y = df['Exited']

# Feature Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

print(X_train.shape)
print(X_test.shape)
```

```

# Build ANN Model

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

model = Sequential()

# Input Layer + Hidden Layer
model.add(Dense(64, activation='relu', input_dim=X_train.shape[1]))
model.add(Dropout(0.3))

# Hidden Layer
model.add(Dense(32, activation='relu'))
model.add(Dropout(0.2))

# Output Layer
model.add(Dense(1, activation='sigmoid'))

# Compile Model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```

# Train Model

history = model.fit(
    X_train,
    y_train,
    validation_data=(X_test, y_test),
    epochs=20,
    batch_size=32
)
```

```

# Model Evaluation

y_pred = model.predict(X_test)
y_pred = (y_pred > 0.5)

print("Accuracy Score:")
print(accuracy_score(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```

# Accuracy Graph

plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')

plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```


# 3. Accident Detection using CNN

## Objective
To detect accidents from images using Convolutional Neural Network (CNN).

## CNN Architecture
1. Convolution Layer
2. ReLU Activation
3. Pooling Layer
4. Flatten Layer
5. Fully Connected Layer


```

# Import CNN Libraries

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
```

```

# Build CNN Model

cnn = Sequential()

# Convolution + Pooling
cnn.add(Conv2D(
    filters=32,
    kernel_size=(3,3),
    activation='relu',
    input_shape=(224,224,3)
))

cnn.add(MaxPooling2D(pool_size=(2,2)))

# Second Convolution Layer
cnn.add(Conv2D(
    filters=64,
    kernel_size=(3,3),
    activation='relu'
))

cnn.add(MaxPooling2D(pool_size=(2,2)))

# Flatten Layer
cnn.add(Flatten())

# Fully Connected Layer
cnn.add(Dense(units=128, activation='relu'))

# Output Layer
cnn.add(Dense(units=1, activation='sigmoid'))

# Compile CNN
cnn.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

cnn.summary()
```


# 4. Padding and Stride in CNN

## Padding
Padding means adding extra pixels around the border of the image.

### Types of Padding
1. Valid Padding → No extra padding  
2. Same Padding → Output size remains same as input size  

### Advantages
- Preserves spatial dimensions
- Prevents information loss at image borders

---

## Stride
Stride defines how many pixels the filter moves during convolution.

### Example
- Stride = 1 → filter moves one pixel at a time
- Stride = 2 → filter skips one pixel

### Effects of Larger Stride
- Reduces output size
- Faster computation
- May lose some information

- Rotation

- Zooming
- Cropping
- Shearing

- Improves model performance
- Enhances generalization capability
